```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df.info()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
 X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
knn.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
knn.score(X_test, y_test)
```

```python
from sklearn.svm import SVC
```

```python
svm = SVC(C=30, kernel='linear', gamma='auto')
```

```python
svm.fit(X_train, y_train)
```

```python
svm.score(X_test, y_test)
```

#**GRID SEARCH CV**

```python
from sklearn.model_selection import GridSearchCV
```

```python
gscv = GridSearchCV((svm), {
    'C': [1,10,20,30],
    'kernel': ['rbf','linear'],
    'gamma': ['auto', 'scale']
}, cv=5, return_train_score=False)
```

```python
gscv.fit(X, y)
```

```python
results = pd.DataFrame(gscv.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel', 'mean_test_score'] ]
```

#**RANDOM SEARCH CV / RANDOMIZED SEARCH**

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
rscv = RandomizedSearchCV((svm), {
    'C': [1,10,20,30],
    'kernel': ['rbf','linear'],
    'gamma': ['auto', 'scale']
},n_iter = 4,cv=5, return_train_score=False)
```

```python
rscv.fit(X, y)
```

```python
results2 = pd.DataFrame(rscv.cv_results_)
```

```python
results2[['param_C','param_kernel', 'mean_test_score'] ]
```
