```python
#STACKING ARRAY
#1) VERTICAL STACK
#2) HORIZONTAL STACK
#3) SIMPLE STACK
```

```python
a = np.array([1,2, 3,4])
b = np.array([5,6, 7,8])
```

```python
#1) VERTICAL STACK
np.vstack((a,b))
```

```python
#2) HORIZONTAL STACK
np.hstack((a,b))
```

```python
#3) SIMPLE STACK
np.column_stack((a,b))
```

```python
#SPLITTING ARRAY
c = np.arange(16).reshape(4,4)
c
```

```python
#HORIZONTAL SPLIT
np.hsplit(c,2)
```

```python
#VERTICAL SPLIT
np.vsplit(c,2)
```
