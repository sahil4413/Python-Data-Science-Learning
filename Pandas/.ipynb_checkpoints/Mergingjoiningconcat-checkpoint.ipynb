```python
import numpy as np 
import pandas as pd
```

```python
employees = pd.DataFrame({
    'employee_id': [1,2,3,4,5],
    'name': ['Sahil', 'Khushi', 'Sanchari', 'Nihal', 'Anik'],
    'department': ['Finance', 'HR', 'IT', 'DS', 'AIML'],
})

salaries = pd.DataFrame({
    'employee_id': [1,2,3,6,7],
    'salary':[60000,80000,65000,85000,75000],
    'bonus': [5000,6000,8000,10000,9000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'inner')
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'outer')
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'left')
```

```python
pd.merge(employees,salaries, on = 'employee_id', how = 'right')
```

**CONCATINATION**

```python

df1 = pd.DataFrame({
    'A': ['A0','A1','A2'],
    'B':['B0','B1','B2'],
    'C':['C0','C1','C2']
})

df2 = pd.DataFrame({
    'A': ['A3','A4','A5'],
    'B':['B3','B4','B5'],
    'C':['C3','C4','C5']
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2])
```

**JOINING 2 DATA FRAMES**

```python
df1 = pd.DataFrame({
    'name':['Alice', 'Bob', 'Charlie']
}, index = [1,2,3])

df2 = pd.DataFrame({
    'score':[85,90,95]
}, index = [2,3,4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2, how = 'outer')
```

```python
df2.join(df1, how = 'outer')
```
